```python
# Импортируем сборщик и справочник категорий.
from yandex_maps_reviews import collect_area, list_categories
# Выбираем территорию из каталога; также доступны bbox и GeoJSON.
city = 'Москва'
# Выбираем категории; полный список доступен через list_categories().
tags = ['cafe', 'residential']
# Получаем до 50 свежих отзывов на организацию; None означает все доступные.
max_reviews = 50
# Сохраняем прогресс отдельно от файлов репозитория.
output = '/content/results_moscow'
# Запускаем поиск и сбор; браузер работает в отдельном процессе.
result = collect_area(city=city, tags=tags, max_orgs=3, max_reviews=max_reviews, output=output)
# Показываем организации, общую оценку и состояние сбора.
display(result.organizations[['organization_id', 'name', 'rating', 'review_status']])
# Показываем первые тексты, звёзды и признаки подтверждения.
display(result.reviews[['organization_id', 'text', 'rating', 'confirmed']].head())
```

## 💾 Сохранить результат
В GeoJSON каждая организация хранится вместе с координатами, оценкой и своими отзывами. Частичный сбор сохраняет исходный статус; `confirmed=False` указывает на неподтверждённую запись. Для анализа подтверждённых полных отзывов используйте `result.confirmed_reviews()`.

Файлы `/content` исчезают при удалении среды Colab. Скачайте результат следующей ячейкой. Для продолжения сбора между разными средами задавайте `output` на предварительно подключённом Google Диске; один GeoJSON не заменяет папку кэша.

```python
# Преобразуем организации в точки WGS84; GeoPandas уже установлен.
geo = result.to_geopandas()
# Сохраняем организации с вложенными отзывами в один GeoJSON.
path = result.save_geojson('/content/organizations_with_reviews.geojson')
# Подключаем штатное скачивание файлов из Colab.
from google.colab import files
# Скачиваем результат на компьютер.
files.download(str(path))
```
